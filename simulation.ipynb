```python
from src import Portfolio, Stocks, Screener, Strategy
```

```python
import pandas as pd

screener = Screener(1e+9, 'br')
screener.screen()

stock_selection = Stocks(screener.local_stocks_names)

for i in range(30): # i for the stoccks to choose, j for the types of portfolio to choose

    set1 = stock_selection.random_selection(5, '7Y')
    values = []
    names = []

    buy_and_hold_strat = Strategy(set1, '7Y', volume_fee=0.005, have_rebalancement=False)
    buy_and_hold_port = Portfolio(buy_and_hold_strat, 'buy_and_hold') 
    buy_and_hold_port.create_port()

    names.append(buy_and_hold_port.name)
    values.append(buy_and_hold_port.sharpe)

    for j in range(6):

        strategy = Strategy(set1, '7Y', volume_fee=0.005, rebalancement_frequency=(j+1)*30)
        port = Portfolio(strategy, f'portfolio_frequency_{(j+1)*30}')
        port.create_port()

    print(f'nemes: {names}')
    print(f'sharpe_values: {values}')

    if i == 0:
        df = pd.DataFrame([dict(zip(names, values))])
    else:
        new_row = pd.DataFrame([values], columns=names)
        df = pd.concat([df, new_row], ignore_index=True)

```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

plt.figure(figsize=(10, 6))

sns.histplot(df, kde=True)

plt.xlim(-1, 2)

plt.title('Probability Distribution of Portfolio Sharpe Ratios')
plt.xlabel('Risk Adjusted Returns')
plt.ylabel('Probability')
plt.show()
```
